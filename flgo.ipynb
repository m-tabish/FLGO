```python
# =======================================================
# ✅ ROBUST FLGO IMPLEMENTATION (CLEANED & OPTIMIZED)
# - Architecture: ResNet-18 (Fast Training)
# - Optimizer: SGD (OADN style)
# - Augmentation: FLGO (Facial Landmark-Guided Occlusion)
# - Fixes: Indentation, Invisible Characters, Pathing
# =======================================================


# --- PART 1: GOOGLE DRIVE SETUP ---
from google.colab import drive
import os


# Mount Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')


# Create save directory
save_dir = "/content/drive/My Drive/AffectNet_Models"
os.makedirs(save_dir, exist_ok=True)
print(f"✅ Models will be saved to: {save_dir}")


# --- PART 2: IMPORTS ---
import glob
import numpy as np
import cv2
import pickle
import copy
import random
from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True  # Handle corrupt images


import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
from torchvision import models
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
from tqdm import tqdm


# --- PART 3: CONFIGURATION ---
print("🚀 Setting up configuration...")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"   Using device: {DEVICE}")


# DATA PATH (Update if needed)
DATA_ROOT = "./affectnet_dataset/YOLO_format"


# Hyperparameters (Optimized for Speed/T4 GPU)
IMG_SIZE = 224
BATCH_SIZE = 128
NUM_CLASSES = 8
OCCLUSION_PROB = 0.5
EPOCHS = 40
PATIENCE = 8


# Optimizer Settings
LEARNING_RATE = 0.001
MOMENTUM = 0.9
WEIGHT_DECAY = 0.0005
SCHEDULER_STEP_SIZE = 7
SCHEDULER_GAMMA = 0.1


# --- PART 4: LOAD LANDMARKS ---
print("🔍 Loading and Normalizing Landmarks...")


# Try local path first, then root path
landmarks_file = os.path.join(DATA_ROOT, "dlib_facial_landmarks.pkl")
if not os.path.exists(landmarks_file):
    landmarks_file = "/content/dlib_facial_landmarks.pkl"


if not os.path.exists(landmarks_file):
    print(f"⚠️ Warning: Landmarks file not found at {landmarks_file}")
    print("   Make sure you uploaded 'dlib_facial_landmarks.pkl' to Colab.")
    # creating dummy dict to prevent crash if file missing during copy-paste testing
    raw_landmarks = {}
else:
    with open(landmarks_file, "rb") as f:
        raw_landmarks = pickle.load(f)


# Normalize keys to filenames only
landmarks_lookup = {}
if raw_landmarks:
    for path_key, lms in raw_landmarks.items():
        filename = os.path.basename(path_key)
        landmarks_lookup[filename] = lms
    print(f"✅ Loaded {len(landmarks_lookup)} landmark entries.")


# --- PART 5: DATASET CLASS ---
class AffectNetDataset(Dataset):
    def __init__(self, img_dir, label_dir, split, occlude_prob=0.0):
        self.img_paths = sorted(glob.glob(os.path.join(img_dir, "*.jpg")))
        self.label_paths = [os.path.join(label_dir, os.path.basename(p).replace(".jpg", ".txt")) for p in self.img_paths]
        self.split = split
        self.occlude_prob = occlude_prob


        # Transforms
        self.normalize = T.Compose([
            T.ToTensor(),
            T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        self.resize = T.Resize((IMG_SIZE, IMG_SIZE))


    def __len__(self):
        return len(self.img_paths)


    def __getitem__(self, idx):
        img_path = self.img_paths[idx]
        filename = os.path.basename(img_path)


        # Load Label
        label = 0
        try:
            with open(self.label_paths[idx], 'r') as f:
                line = f.readline().strip()
                if line: label = int(line.split()[0])
        except:
            pass


        # Load Image
        try:
            image = Image.open(img_path).convert("RGB")
        except:
            # Create black image if load fails
            image = Image.new('RGB', (IMG_SIZE, IMG_SIZE))

        w_orig, h_orig = image.size
        img_arr = np.array(image)


        # --- FLGO AUGMENTATION ---
        if self.split == 'train' and self.occlude_prob > 0 and random.random() < self.occlude_prob:
            lms = landmarks_lookup.get(filename)

            # Draw occlusion if landmarks exist
            if lms is not None and len(lms) == 68:
                regions = {
                    "mouth": list(range(48, 68)),
                    "left_eye": list(range(36, 42)),
                    "right_eye": list(range(42, 48)),
                    "nose": list(range(27, 36))
                }

                region_name = random.choice(list(regions.keys()))
                points = [lms[i] for i in regions[region_name]]

                if len(points) > 0:
                    xs = [p[0] for p in points]
                    ys = [p[1] for p in points]

                    minx, maxx = max(0, min(xs)), min(w_orig, max(xs))
                    miny, maxy = max(0, min(ys)), min(h_orig, max(ys))

                    if maxx > minx and maxy > miny:
                        img_arr[miny:maxy, minx:maxx, :] = 0

            image = Image.fromarray(img_arr)


        # Resize & Normalize
        image = self.resize(image)

        # Horizontal Flip
        if self.split == 'train' and random.random() < 0.5:
            image = T.functional.hflip(image)

        return self.normalize(image), label


# --- PART 6: VISUAL DEBUGGER ---
def debug_visualize(dataset, count=5):
    print("\n📸 DEBUG: Visualizing Augmented Samples...")
    fig, axes = plt.subplots(1, count, figsize=(15, 4))

    original_prob = dataset.occlude_prob
    dataset.occlude_prob = 1.0 # Force occlusion

    valid_samples = 0
    idx = 0
    while valid_samples < count and idx < len(dataset):
        try:
            img_tensor, label = dataset[idx]
            # Undo normalization
            img = img_tensor.permute(1, 2, 0).numpy()
            img = img * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406])
            img = np.clip(img, 0, 1)

            axes[valid_samples].imshow(img)
            axes[valid_samples].set_title(f"Label: {label}")
            axes[valid_samples].axis('off')
            valid_samples += 1
        except:
            pass
        idx += 1

    dataset.occlude_prob = original_prob
    plt.tight_layout()
    plt.show()
    print("✅ Check images above for black boxes.")


# --- PART 7: MODEL & TRAINING LOOP ---
def get_model():
    model = models.resnet18(pretrained=True)
    model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)
    return model.to(DEVICE)


def run_training(loader, val_loader, model_name):
    model = get_model()
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.SGD(model.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM, weight_decay=WEIGHT_DECAY)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='max',      # We want to MAXIMIZE accuracy
    factor=0.1,      # Reduce LR by 10x when stuck
    patience=3,      # Wait 3 epochs before reducing LR
         # Print message when LR changes
)

    best_acc = 0.0
    patience_counter = 0
    train_losses, val_accs = [], []

    print(f"\n🚀 STARTING TRAINING: {model_name}")

    for epoch in range(EPOCHS):
        # Train
        model.train()
        running_loss = 0.0
        for imgs, lbls in tqdm(loader, desc=f"Ep {epoch+1} Train"):
            imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, lbls)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()

        avg_train_loss = running_loss / len(loader)
        train_losses.append(avg_train_loss)

        # Validate
        model.eval()
        all_preds, all_lbls = [], []
        with torch.no_grad():
            for imgs, lbls in val_loader:
                imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
                outputs = model(imgs)
                preds = torch.argmax(outputs, 1)
                all_preds.extend(preds.cpu().numpy())
                all_lbls.extend(lbls.cpu().numpy())

        val_acc = accuracy_score(all_lbls, all_preds)
        val_accs.append(val_acc)

        # Step Scheduler & Get Current LR
        scheduler.step(val_acc)
        current_lr = optimizer.param_groups[0]['lr']

        # --- PRINT ITERATION STATS ---
        print(f"   [Ep {epoch+1}/{EPOCHS}] {model_name} | LR: {current_lr:.6f} | Loss: {avg_train_loss:.4f} | Val Acc: {val_acc:.4f}")

        # Save Best & Early Stopping
        if val_acc > best_acc:
            best_acc = val_acc
            patience_counter = 0
            save_path = os.path.join(save_dir, f"best_{model_name}.pth")
            torch.save(model.state_dict(), save_path)
            print(f"   💾 Saved Best Model to: {save_path}")
        else:
            patience_counter += 1
            if patience_counter >= PATIENCE:
                print(f"🛑 Early Stopping at Epoch {epoch+1}")
                break

    return best_acc, train_losses, val_accs


# --- PART 8: EXECUTION ---
def main():
    # 1. Setup Directories
    train_dir = os.path.join(DATA_ROOT, "train", "images")
    train_lbl = os.path.join(DATA_ROOT, "train", "labels")
    val_dir = os.path.join(DATA_ROOT, "valid", "images")
    val_lbl = os.path.join(DATA_ROOT, "valid", "labels")


    if not os.path.exists(train_dir):
        print(f"❌ Error: Data directory not found at {train_dir}")
        return


    # 2. Datasets
    print("📂 Creating Datasets...")
    train_ds_base = AffectNetDataset(train_dir, train_lbl, 'train', occlude_prob=0.0)
    train_ds_flgo = AffectNetDataset(train_dir, train_lbl, 'train', occlude_prob=OCCLUSION_PROB)
    val_ds = AffectNetDataset(val_dir, val_lbl, 'valid', occlude_prob=0.0)


    # 3. Visual Check
    try:
        debug_visualize(train_ds_flgo)
    except Exception as e:
        print(f"⚠️ Visualization skipped: {e}")


    # 4. DataLoaders
    train_loader_base = DataLoader(train_ds_base, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
    train_loader_flgo = DataLoader(train_ds_flgo, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)


    # 5. Run Training
    print("\n--- Phase 1: Baseline (No Occlusion) ---")
    base_acc, base_loss, base_val = run_training(train_loader_base, val_loader, "Baseline_ResNet18")


    print("\n--- Phase 2: FLGO (With Occlusion) ---")
    flgo_acc, flgo_loss, flgo_val = run_training(train_loader_flgo, val_loader, "FLGO_ResNet18")


    # 6. Results
    print("\n" + "="*50)
    print(f"🏆 FINAL RESULTS")
    print(f"Baseline Accuracy: {base_acc:.4f}")
    print(f"FLGO Accuracy:     {flgo_acc:.4f}")
    print(f"Improvement:       {(flgo_acc - base_acc)*100:.2f}%")
    print("="*50)


    # Plot
    plt.figure(figsize=(10, 5))
    plt.plot(base_val, label=f'Baseline (Best: {base_acc:.2f})')
    plt.plot(flgo_val, label=f'FLGO (Best: {flgo_acc:.2f})')
    plt.title("Validation Accuracy")
    plt.xlabel("Epochs")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.savefig("final_comparison.png")
    print("Plot saved as final_comparison.png")


if __name__ == "__main__":
    main()
```

```python
# --- ROBUSTNESS EVALUATION SCRIPT ---
def evaluate_robustness(model, loader, device, model_name):
    model.eval()
    all_preds, all_lbls = [], []
    with torch.no_grad():
        for imgs, lbls in tqdm(loader, desc=f"Testing {model_name}"):
            imgs, lbls = imgs.to(device), lbls.to(device)
            outputs = model(imgs)
            preds = torch.argmax(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_lbls.extend(lbls.cpu().numpy())
    return accuracy_score(all_lbls, all_preds)

# 1. Load your best saved models
print("🔄 Loading Models for Robustness Test...")
model_base = get_model() # ResNet18
model_base.load_state_dict(torch.load(f"{save_dir}/best_Baseline_ResNet18.pth"))
model_base.eval()

model_flgo = get_model() # ResNet18
model_flgo.load_state_dict(torch.load(f"{save_dir}/best_FLGO_ResNet18.pth"))
model_flgo.eval()

# 2. Create an OCCLUDED Validation Set (Force Occlusion = 1.0)
print("📂 Creating Occluded Test Set...")
val_dir = os.path.join(DATA_ROOT, "valid", "images")
val_lbl = os.path.join(DATA_ROOT, "valid", "labels")

# Note: We use 'train' split logic here just to enable the augmentation,
# but we point it to the VALIDATION folders.
test_ds_clean = AffectNetDataset(val_dir, val_lbl, 'valid', occlude_prob=0.0)
test_ds_occluded = AffectNetDataset(val_dir, val_lbl, 'train', occlude_prob=1.0) # Force 100% Occlusion

test_loader_clean = DataLoader(test_ds_clean, batch_size=BATCH_SIZE, shuffle=False)
test_loader_occ = DataLoader(test_ds_occluded, batch_size=BATCH_SIZE, shuffle=False)

# 3. Run Comparisons
print("\n--- 1. CLEAN VALIDATION SET ---")
acc_base_clean = evaluate_robustness(model_base, test_loader_clean, DEVICE, "Baseline")
acc_flgo_clean = evaluate_robustness(model_flgo, test_loader_clean, DEVICE, "FLGO")

print("\n--- 2. OCCLUDED VALIDATION SET (The Real Test) ---")
acc_base_occ = evaluate_robustness(model_base, test_loader_occ, DEVICE, "Baseline")
acc_flgo_occ = evaluate_robustness(model_flgo, test_loader_occ, DEVICE, "FLGO")

# 4. Print The "Winning" Table
print("\n" + "="*50)
print("🏆 ROBUSTNESS ANALYSIS")
print(f"{'Metric':<20} | {'Baseline':<10} | {'FLGO (Yours)':<10} | {'Gap'}")
print("-" * 50)
print(f"{'Clean Accuracy':<20} | {acc_base_clean:.4f}     | {acc_flgo_clean:.4f}       | {acc_flgo_clean-acc_base_clean:.4f}")
print(f"{'Occluded Accuracy':<20} | {acc_base_occ:.4f}     | {acc_flgo_occ:.4f}       | +{acc_flgo_occ-acc_base_occ:.4f}")
print("-" * 50)

if acc_flgo_occ > acc_base_occ:
    print("✅ CONCLUSION: FLGO is significantly more robust to occlusion.")
else:
    print("❌ CONCLUSION: FLGO needs more training time.")
```

```python
# Visualisation
import matplotlib.pyplot as plt
import numpy as np

# Your Data
metrics = ['Clean Accuracy', 'Occluded Accuracy']
baseline_scores = [64.16, 56.01]
flgo_scores = [63.51, 60.29]

x = np.arange(len(metrics))  # Label locations
width = 0.35  # Bar width

fig, ax = plt.subplots(figsize=(8, 6))
rects1 = ax.bar(x - width/2, baseline_scores, width, label='Baseline', color='#d62728') # Red
rects2 = ax.bar(x + width/2, flgo_scores, width, label='FLGO (Ours)', color='#2ca02c') # Green

# Add text for labels, title and custom x-axis tick labels, etc.
ax.set_ylabel('Accuracy (%)')
ax.set_title('Robustness Analysis: Baseline vs. FLGO')
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.set_ylim(50, 70) # Zoom in to see the difference
ax.legend()

# Add labels on top of bars
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}%',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)

fig.tight_layout()
plt.savefig("robustness_chart.png")
plt.show()

from sklearn.metrics import classification_report, f1_score, precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay

def run_comprehensive_eval(model, loader, device, model_name, class_names=None):
    model.eval()
    all_preds, all_lbls = [], []

    # 1. Gather Predictions
    with torch.no_grad():
        for imgs, lbls in tqdm(loader, desc=f"Evaluating {model_name}"):
            imgs, lbls = imgs.to(device), lbls.to(device)
            outputs = model(imgs)
            preds = torch.argmax(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_lbls.extend(lbls.cpu().numpy())

    # 2. Calculate Metrics
    acc = accuracy_score(all_lbls, all_preds)
    # 'weighted' accounts for class imbalance (important for AffectNet)
    f1 = f1_score(all_lbls, all_preds, average='weighted')
    recall = recall_score(all_lbls, all_preds, average='weighted')
    precision = precision_score(all_lbls, all_preds, average='weighted')

    print(f"\n" + "="*40)
    print(f"📊 REPORT: {model_name}")
    print(f"="*40)
    print(f"Accuracy:  {acc:.4f}")
    print(f"F1 Score:  {f1:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"Precision: {precision:.4f}")

    # 3. Detailed Per-Class Report
    if class_names is None:
        class_names = [str(i) for i in range(8)] # Default 0-7

    print("\n--- Per-Class Metrics ---")
    print(classification_report(all_lbls, all_preds, target_names=class_names, digits=4))

    # 4. Plot Confusion Matrix
    cm = confusion_matrix(all_lbls, all_preds)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)

    fig, ax = plt.subplots(figsize=(8, 8))
    disp.plot(cmap='Blues', ax=ax, colorbar=False)
    plt.title(f"Confusion Matrix: {model_name}")
    plt.show()

    return acc, f1

# --- USAGE EXAMPLE ---
# Define AffectNet Classes (Standard 8 expressions)
classes = ['Neutral', 'Happy', 'Sad', 'Surprise', 'Fear', 'Disgust', 'Anger', 'Contempt']

print("Running Final Detailed Evaluation...")

# Evaluate Baseline
run_comprehensive_eval(model_base, test_loader_occ, DEVICE, "Baseline (Occluded)", classes)

# Evaluate FLGO
run_comprehensive_eval(model_flgo, test_loader_occ, DEVICE, "FLGO (Occluded)", classes)
```
